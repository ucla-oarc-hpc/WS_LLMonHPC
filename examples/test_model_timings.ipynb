```python
import requests
import json
from pprint import pprint
```

```python
# Configuration
OLLAMA_API_URL = "http://localhost:11434/api"
```

```python
def get_installed_models():
    """
    Fetch the list of installed models from the Ollama service.
    """
    endpoint = f"{OLLAMA_API_URL}/tags"
    try:
        response = requests.get(endpoint)
        response.raise_for_status()  # Raise an error for bad status codes
        models = response.json()
        # Pretty print the JSON output
        print("Installed Models:")
        print(json.dumps(models, indent=4))
        return None
    except requests.exceptions.RequestException as e:
        print("Error fetching models:", e)
        return None
```

```python
get_installed_models()
```

```python
# Create a list of models to test
models_to_test = [
    "neural-chat:latest",
    "llama3.2:3b"
]
```

```python
import time
model_timings = {}
def generate_text(prompt, model="llama3", stream=False):
    """Generate text using a model and display the formatted response."""
    data = {
        "model": model,
        "prompt": prompt,
        "stream": stream
    }
    
    try:
        response = requests.post(f"{OLLAMA_API_URL}/generate", json=data)
        response.raise_for_status()
        result = response.json()
        
        print("\nGenerated Response:")
        pprint(result['response'], indent=2)
    except Exception as e:
        print(f"Error generating text: {e}")
```

```python
for model_name in models_to_test:
    print(f"\nTesting model: {model_name}")
    start = time.time()
    generate_text("Explain AI in simple terms.", model=model_name)
    end = time.time()
    wall_time = end - start
    
    model_timings[model_name] = wall_time
    print(f"Wall time for running {model_name} is: {wall_time}")
    print("-" * 50) 
```

```python
print(model_timings)
```

```python
import matplotlib.pyplot as plt

models = list(model_timings.keys())
wall_times = list(model_timings.values())

plt.figure(figsize=(10, 6))
plt.bar(models, wall_times)
plt.xlabel('Quantized Model (Qwen is Q8, Rest are Q4)')
plt.ylabel('Wall Time (seconds)')
plt.title('Wall Time Comparison for Different Models')
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
