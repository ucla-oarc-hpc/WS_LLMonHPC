```python
import requests
import json
from IPython.display import display, Markdown

# Ollama settings
OLLAMA_URL = "http://localhost:11434/api/chat"
DEFAULT_MODEL = "llama3.2:3B"

messages = [
    {"role": "system", "content": "You are a helpful assistant who love puns "}
]

def chat_with_ollama(user_input, model=DEFAULT_MODEL):
    """
    Send a single message to the chat model and return/display the response.
    """
    messages.append({"role": "user", "content": user_input})

    payload = {
        "model": model,
        "messages": messages,
        "stream": True
    }

    assistant_message = ""
    try:
        response = requests.post(OLLAMA_URL, json=payload, stream=True)

        for line in response.iter_lines():
            if line:
                data = json.loads(line)
                delta = data.get("message", {}).get("content", "")
                assistant_message += delta

        messages.append({"role": "assistant", "content": assistant_message})

        # Display nicely in Jupyter
        display(Markdown(f"**AI:** {assistant_message}"))

    except Exception as e:
        display(Markdown(f"**[Error]** {e}"))
```

```python
chat_with_ollama("Create python code for a Linear Regression example")
```

```python
chat_with_ollama("Rewrite the code in R")
```
